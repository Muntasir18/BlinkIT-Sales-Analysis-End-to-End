```python
import pandas as pd
from numpy.ma.extras import unique
```

```python
data = pd.read_csv('BlinkIT_Grocery_Data.csv', encoding='utf-8-sig', sep=';')
data = pd.DataFrame(data)
data
```

```python
data['Item Fat Content'].unique()
```

```python
data['Item Fat Content'] = data['Item Fat Content'].replace({
    'LF': 'Low Fat',
    'reg': 'Regular',
    'low fat': 'Low Fat'
})
```

```python
data['Item Fat Content'].unique()
```

```python
data.columns = [c.replace(' ', '_') for c in data.columns]
data
```

```python
df = data.where(pd.notnull(data), None)
```

```python
df.to_csv(
    "/Users/callachand.muntasir/Data_Analyst_projet/Blinkit_report/BlinkIT_Grocery_Data_clean.csv",
    index=False,
    encoding='utf-8',
    na_rep='\\N'
)
```
